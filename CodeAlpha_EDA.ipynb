```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/CodeAlpha_Task1_WebScraping.csv")
print(df)
```

```python
print(df.info())
```

```python
df["Price"] = df["Price"].str.replace("£", "", regex=False).astype(float)

print(df)
```

```python
print(df["Price"].describe())
```

```python
print(df["Rating"].value_counts())
```

```python
df["Rating"].value_counts().plot(kind="bar")

plt.title("Book Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.show()
```

```python
plt.hist(df["Price"], bins=5)

plt.title("Book Price Distribution")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.show()
```

```python
rating_order = ["One", "Two", "Three", "Four", "Five"]

sns.boxplot(x="Rating", y="Price", data=df, order=rating_order)

plt.title("Book Price by Rating")
plt.xlabel("Rating")
plt.ylabel("Price (£)")
plt.show()
```

```python
print("EDA Summary")
print("--------------------")
print("Total Books:", len(df))
print("Average Price: £", round(df["Price"].mean(), 2))
print("Minimum Price: £", round(df["Price"].min(), 2))
print("Maximum Price: £", round(df["Price"].max(), 2))
print("Most Common Rating:", df["Rating"].value_counts().idxmax())
```

```python
df.to_csv("CodeAlpha_Task2_EDA.csv", index=False)
print("EDA dataset saved successfully!")
```
